```python
from sentence_transformers import SentenceTransformer

# Fixed model name
model = SentenceTransformer("all-MiniLM-L6-v2")  # Fast and tiny (384 dimensions)

quotes = [
    "To be, or not to be, that is the question.",
    "May the Force be with you.",
    "Houston, we have a problem.",
    "I'm going to make him an offer he can't refuse.",
]

embeddings = model.encode(quotes)
print(embeddings.shape)  # Should output (4, 384)
```

```python
from core.index import HNSWIndex  # Your custom database

db = HNSWIndex(dim=384, M=16)
for i, vector in enumerate(embeddings):
  db.insert(vector=vector, metadata={"quote": quotes[i]})
```

```python
%pip install streamlit
```

```python
import streamlit as st

st.title("Semantic Quote Finder 🔍")
query = st.text_input("How are you feeling or what are you looking for?")

if query:
  # 1. Convert user query to vector
  query_vector = model.encode(query)

  # 2. Search using your custom HNSW engine!
  results = db.search(query_vector, efSearch=32, k=1)

  # 3. Display the result
  st.success(f"Best Match: {results[0]['metadata']['quote']}")
```
